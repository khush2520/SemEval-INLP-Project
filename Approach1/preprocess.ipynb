```python
def read_ann_file(filename):
    annotations = []

    with open(filename, 'r') as f:
        for line in f:

            if not line:
                continue
    
            l = line.split('\t')

            if line[0] != 'T':
                continue
            id, labels, surface_form = l
            
            label, start_offset, end_offset = labels.split(' ')
            line2 = id + '\t' + label + ' ' + start_offset + ' ' + end_offset + '\t' + surface_form
            annotations.append(line2)
    return annotations

# filename = os.path.join('../train2', 'S0003491615000433.ann')
# annotations = read_ann_file(filename)
# for annotation in annotations:
#   print(annotation)
```

```python

import os
import pandas as pd

folder_path = os.path.join('', '../semeval_articles_test')
output_folder = os.path.join('', 'test_witout_task3_labels')

df = pd.DataFrame(columns=['file_name', 'text', 'annotations'])

for file_name in os.listdir(folder_path):
  if file_name.endswith('.txt'):
    text_file_path = os.path.join(folder_path, file_name)

    with open(text_file_path, 'r') as f:
      text = f.read()

    ann_file_path = os.path.join(folder_path, file_name[0:-4] + '.ann')

    annotations = read_ann_file(ann_file_path)

    output_file_path = os.path.join(output_folder, file_name[0:-4] + '.ann')
    with open(output_file_path, 'w') as f:
      for annotation in annotations:
        f.write(annotation)
```
